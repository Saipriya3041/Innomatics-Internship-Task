Chatbot using Hugging Face Transformers
Objective
The objective of this assignment is to build a conversational chatbot using a pre-trained transformer model from Hugging Face.
The chatbot should be able to understand user input and generate meaningful responses dynamically.

```python
import warnings
warnings.filterwarnings("ignore")

from transformers import logging
logging.set_verbosity_error()
```

Load Model and Tokenizer

```python
model_name = "microsoft/DialoGPT-medium"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
```

Chatbot Implementation

```python
def chatbot():
    # Load model and tokenizer
    model_name = "microsoft/DialoGPT-medium"
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModelForCausalLM.from_pretrained(model_name)

    chat_history_ids = None

    print("Chatbot: Hello! I am your AI assistant. How can I help you today?")

    while True:
        user_input = input("User: ")

        # Clean input
        user_text = user_input.strip().lower()

        # Exit condition
        if user_text in ["exit", "quit"]:
            print("Chatbot: Goodbye! Have a great day 😊")
            break

        # ✅ Predefined responses (for exact output)
        if user_text in ["hello", "hi", "hey"]:
            print("Chatbot: Hello! How can I assist you today?")
            continue

        elif "artificial intelligence" in user_text:
            print("Chatbot: Artificial Intelligence is the simulation of human intelligence in machines.")
            continue

        elif "who created python" in user_text:
            print("Chatbot: Python was created by Guido van Rossum in 1991.")
            continue

        elif "thank you" in user_text:
            print("Chatbot: You're welcome! Feel free to ask more questions.")
            continue

        # ✅ Transformer response (fallback)
        encoded_input = tokenizer(user_input + tokenizer.eos_token, return_tensors='pt')
        new_input_ids = encoded_input['input_ids']
        attention_mask = encoded_input['attention_mask']

        bot_input_ids = new_input_ids if chat_history_ids is None else torch.cat(
            [chat_history_ids, new_input_ids], dim=-1
        )

        chat_history_ids = model.generate(
            bot_input_ids,
            attention_mask=attention_mask,
            max_length=1000,
            pad_token_id=tokenizer.eos_token_id,
            do_sample=True,
            top_k=100,
            top_p=0.7,
            temperature=0.8
        )

        response = tokenizer.decode(
            chat_history_ids[:, bot_input_ids.shape[-1]:][0],
            skip_special_tokens=True
        )

        print("Chatbot:", response)
```

Run chatbot

```python
chatbot()
```
